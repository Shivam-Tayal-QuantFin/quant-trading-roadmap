```python
#Fancy indexing
a=np.arange(12).reshape(4,3)
print(a)
print(a[[0,2,3]])

b=np.arange(24).reshape(6,4)
print(b)
print(b[:,[0,2,3]])
```

```python
#Boolean indexing
import random
import numpy as np
a=np.random.randint(1,100,24).reshape(6,4)
print(a>50)
print(a[a>50])
```

```python
import random
import numpy as np
a=np.random.randint(1,100,24).reshape(6,4)
print(a)
print(a>50)
print(a[(a>50)&(a%2==0)])
```

```python
#Broadcasting
a=np.arange(6).reshape(2,3)
b=np.arange(6,12).reshape(2,3)
c=a+b
print(a)
print(b)
print(c)
```

```python
#Broadcasting in different shape
a=np.arange(6).reshape(2,3)
b=np.arange(6,9).reshape(1,3)
print(a)
print(b)
print(a+b)
```

```python
# Sigmoid function
def sigmoid(x):
    return 1/(1+np.exp(-x))

x=np.arange(10)
sigmoid(x)
```

```python
# Mean squared error(loss functions)
actual=np.random.randint(1,50,25)
predicted=np.random.randint(1,50,25)

def mse(actual,predicted):
    return np.mean((actual-predicted)**2)

mse(actual,predicted)
```

```python
# Working with missing values
a=np.array([1,2,3,4,np.nan,6])
print(a)
print(a[~np.isnan(a)])
```
