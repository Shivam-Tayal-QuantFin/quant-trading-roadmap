```python
#Calculate the final value of an investment after a specified number of years.
import math
import numpy as np
print("Welcome to compound interest calculator")
a=float(input("Enter your investment amount : "))
b=float(input("Enter the rate of interest : "))
c=float(input("Enter the time of maturity in years : "))
d=a*((1+0.01*b)**c)
e=d-a
print("Profit earned = ",e)
print("Total amount at maturity = ",d)
```

```python
#A trader buys a stock for ₹250 and sells it for ₹275.
#Write a program that calculates:
#Profit or loss.
#Percentage profit or loss.
#Final result: profit, loss, or break-even.
#Your challenge: Add brokerage charges to calculate net profit or loss.
cp=250
sp=275
margin=sp-cp
print("Buying price : ",cp)
print("Selling price : ",sp)
br_per=float(input("Enter the brokerage charges in % : "))
net=margin-(cp*(br_per/100))
net_per=(net/cp)*100
if net>0:
    print("You make a profit of ",net)
    print("Profit : ",net_per,"%")
elif net==0:
    print("You have break-even")
    print("Break even- no loss no profit")
else:
    print("You got a loss of ",net*-1)
    print("Loss : ",net_per*-1,"%")

    
```

```python
#Portfolio Value Calculator
import numpy as np 
import math
arr1=np.array(list(map(int,input("Enter the stock prices : ").split())))
arr2=np.array(list(map(int,input("Enter the number of stocks : ").split())))
a=np.size(arr1)
b=np.size(arr2)
total=0
if a==b:
    for i in range(len(arr1)):
        port=arr1[i]*arr2[i]
        print("Portfolio of company ",i+1,":",port)
        total=total+port
    print("Total portfolio : ",total)

    for i in range(len(arr1)):
        contri=arr1[i]*arr2[i]
        print("Portfolio contribution of company ",i+1," is",(contri/total)*100,"%")
else:
    print("False information")
```

```python
#Compare Two Stocks
#Write a program that:
#Calculates the total price change of each stock.
#Calculates each stock's percentage return.
#Identifies which stock had the higher return.
#Important: Higher return does not necessarily mean lower risk. You will study that distinction later.
import numpy as np
import math
arr1=np.array(list(map(int,input("Enter the stock 1 values : ").split())))
arr2=np.array(list(map(int,input("Enter the stock 2 values : ").split())))
a=np.size(arr1)
b=np.size(arr2)
if a!=b:
    print("False information")
else:
    ret1=arr1[-1]-arr1[0]
    ret2=arr2[-1]-arr2[0]
    print("Price change of stock 1:",ret1)
    print("Price change of stock 2:",ret2)
    pr1=(ret1/arr1[0])*100
    pr2=(ret2/arr2[0])*100
    print("Return by stock 1:",pr1,"%")
    print("Return by stock 2:",pr2,"%")

    if pr1>pr2:
        print("Stock 1 gives",pr1-pr2,"% more return than Stock 2")
    elif pr1==pr2:
        print("Both stocks gave same returns")
    else:
         print("Stock 2 gives",pr2-pr1,"% more return than Stock 1")


        
```

```python
#Write a program to:
#Sort the prices in ascending order.
#Sort them in descending order.
#Find the three highest prices.
import numpy as np 
import math
stock=np.array(list(map(int,input("Enter the stock prizes : ").split())))
a=np.sort(stock)
print("Stock prices in ascending order : ",a)
b=np.flip(a)
print("Stock prices in descending order : ",b)
print("Three highest prices are :",b[0:3])
```

```python
# Compare two portfolios
#Find:The common elements.
#All distinct elements across both portfolios.
#Elements present in portfolio A but not portfolio B.
import numpy as np
import math
port1=np.array(list(map(int,input("The the elements of Portfolio A: ").split())))
port2=np.array(list(map(int,input("The the elements of Portfolio B: ").split())))

a=np.union1d(port1,port2)
b=np.intersect1d(port1,port2)
c=np.setdiff1d(port1,port2)
print("Common elements in both portfolios are :",b)
print("All distinct elements across both portfolios are :",a)
print("Elements present in portfolio A but not portfolio B are :",c)
```

```python
#Combine Financial Datasets
stock1=np.random.randint(225,250,24).reshape(6,4)
print(stock1)
stock2=np.random.randint(225,250,24).reshape(6,4)
print(stock2)
stock3=np.concatenate((stock1,stock2),axis=1)
print("After vertical concatenation :")
print(stock3)
stock6=np.concatenate((stock1,stock2),axis=0)
print("After horizontal concatenation :")
print(stock6)
stock4=np.vstack((stock1,stock2))
print("After vertical stacking :")
print(stock4)
stock5=np.hstack((stock1,stock2))
print("After horizontal stacking :")
print(stock5)


```
