```python
# Numpy arrays
#1D arrays
import numpy as np
a=np.array([1,2,3])
print(a)
print(type(a))
```

```python
# 2D arrays
import numpy as np
a=np.array([[1,2],[3,4]])
print(a)
print(type(a))
```

```python
# 3D arrays
import numpy as np
a=np.array([[[1,2],[3,4]],[[5,6],[7,8]]])
print(a)
print(type(a))
```

```python
# Change datatype of an array 
import numpy as np 
a=np.array([1,2,3],dtype=float)
print(a)
print(type(a))
b=np.array([1,2,3],dtype=bool)
print(b)
print(type(b))
c=np.array([1,2,3],dtype=complex)
print(c)
print(type(c))
```

```python
# Creating arrays using np.arange
a=np.arange(1,11)
b=np.arange(1,11,2)
print(a)
print(type(a))
print(b)
print(type(b))
```

```python
# Reshape arrays whoch are maked by np.arange
a=np.arange(1,11).reshape(2,5)
print(a)
print(type(a))
b=np.arange(0,12).reshape(4,3)
print(b)
print(type(b))
```

```python
# np.ones and np.zeros
a=np.ones((3,4))
print(a)
print(type(a))
b=np.zeros((2,8))
print(b)
print(type(b))
```

```python
# Random value in arrays
a=np.random.random((3,4))
print(a)
print(type(a))
```

```python
# Create linear points on equal distance linearly
a=np.linspace(-10,10,10)
print(a)
print(type(a))
```

```python
# creating an identity matrix
a=np.identity(3)
print(a)
print(type(a))
```

```python
# ndim dimensions of array
a1=np.arange(10)
a2=np.arange(12,dtype=float).reshape(3,4)
a3=np.arange(8).reshape(2,2,2)
print(a1.ndim)
print(a2.ndim)
print(a3.ndim)
```

```python
#shape tells the shape of array
a1=np.arange(10)
a2=np.arange(12,dtype=float).reshape(3,4)
a3=np.arange(8).reshape(2,2,2)
print(a1.shape)
print(a2.shape)
print(a3.shape)

```

```python
# Tells size of the array
a1=np.arange(10)
a2=np.arange(12,dtype=float).reshape(3,4)
a3=np.arange(8).reshape(2,2,2)
print(a1.size)
print(a2.size)
print(a3.size)
```

```python
# itemsize tells the size of the datatype in item
a1=np.arange(10)
a2=np.arange(12,dtype=float).reshape(3,4)
a3=np.arange(8).reshape(2,2,2)
print(a1.itemsize)
print(a2.itemsize)
print(a3.itemsize)
```

```python
#dtype , tells datatype of the item
a1=np.arange(10)
a2=np.arange(12,dtype=float).reshape(3,4)
a3=np.arange(8).reshape(2,2,2)
print(a1.dtype)
print(a2.dtype)
print(a3.dtype)
```

```python
# Changing datatype
a1=np.arange(10)
a2=np.arange(12,dtype=float).reshape(3,4)
a3=np.arange(8).reshape(2,2,2)
print(a3.astype(np.int32))
```

```python
#scalar operations
a1=np.arange(12).reshape(3,4)
a2=np.arange(12,24).reshape(3,4)
print(a1+2)
print(a1-2)
print(a1*2)
print(a1/2)
print(a1**2)
print(a1%2)
print(a2+2)
print(a2-2)
print(a2*2)
print(a2/2)
print(a2**2)
print(a2%2)
```

```python
#vector operations
a1=np.arange(12).reshape(3,4)
a2=np.arange(12,24).reshape(3,4)
print(a1+a2)
print(a1-a2)
print(a1*a2)
print(a1/a2)
print(a1**a2)
print(a1%a2)
```

```python
#Array functions
a1=np.random.random((3,3))
a2=np.round(a1*100)
#sum of elements
print(np.sum(a2))
#product of elements
print(np.prod(a2))
#maximum of element
print(np.max(a2))
#minimum of elements
print(np.min(a2))
#mean of elements
print(np.mean(a2))
#median of elements
print(np.median(a2))
#variance of elements
print(np.var(a2))
#standard deviation of elements
print(np.std(a2))
```

```python
#Array functions rowwise
a1=np.random.random((3,3))
a2=np.round(a1*100)
#sum of elements
print(np.sum(a2,axis=1))
#product of elements
print(np.prod(a2,axis=1))
#maximum of element
print(np.max(a2,axis=1))
#minimum of elements
print(np.min(a2,axis=1))
#mean of elements
print(np.mean(a2,axis=1))
#median of elements
print(np.median(a2,axis=1))
#variance of elements
print(np.var(a2,axis=1))
#standard deviation of elements
print(np.std(a2,axis=1))
```

```python
#Array functions columnwise
a1=np.random.random((3,3))
a2=np.round(a1*100)
#sum of elements
print(np.sum(a2,axis=0))
#product of elements
print(np.prod(a2,axis=0))
#maximum of element
print(np.max(a2,axis=0))
#minimum of elements
print(np.min(a2,axis=0))
#mean of elements
print(np.mean(a2,axis=0))
#median of elements
print(np.median(a2,axis=0))
#variance of elements
print(np.var(a2,axis=0))
#standard deviation of elements
print(np.std(a2,axis=0))
```

```python
#Trigonometric functions
a1=np.random.random((3,3))
a2=np.round(a1*100)
print(np.sin(a2))
print(np.cos(a2))
print(np.tan(a2))
```

```python
# Dot product 
a1=np.arange(12).reshape(3,4)
a3=np.arange(12,24).reshape(4,3)
print(np.dot(a1,a3))
```

```python
#Logarithms and exponents
a1=np.arange(1,13).reshape(3,4)
a3=np.arange(12,24).reshape(4,3)
print(np.log(a1))
print(np.log(a3))
print(np.exp(a1))
print(np.exp(a3))
```

```python
#Round floor and ceil 
a1=(np.random.random((2,3))*100)
print(np.round(a1))
print(np.floor(a1))
print(np.ceil(a1))
```

```python
#Indexing and slicing 
import numpy as np
a1=np.arange(10)
a2=np.arange(12).reshape(3,4)
a3=np.arange(8).reshape(2,2,2)
print(a1[1])
print(a3[1,0,1])
print(a3[1,1,0])
print(a1[2:5])
print(a2[0,:])
print(a2[:,2])
print(a2[::2,::3])
print(a2[::2,1::2])
print(a2[1,::3])
print(a2[0:2,1:])
a4=np.arange(27).reshape(3,3,3)
print(a4[2,1:,1:])
print(a4[0::2,0,0::2])
```

```python
#Iterating
import numpy as np
a3=np.arange(8).reshape(2,2,2)
for i in np.nditer(a3):
    print(i)

a4=np.random.random((6,5))
a5=np.round(a4*100)
for j in np.nditer(a5):
    print(j)
```

```python
#Transpose of a matrix
import numpy as np
a1=np.random.random((4,6))
a2=np.round(100*a1)
print(a2)
a3=np.transpose(a2)
print(a3)
```

```python
# Ravel- Convert X-d array into 1D array
a1=np.arange(12).reshape(2,3,2)
print(a1)
print(a1.ravel())
```

```python
#horizontal stacking -> adding of arrays horizontally
a1=np.arange(12).reshape(3,4)
a2=np.arange(12,24).reshape(3,4)
a3=np.hstack((a1,a2))
print(a1)
print(a2)
print(a3)
```

```python
#vertical stacking ->adding of arrays vertically
a1=np.arange(12).reshape(3,4)
a2=np.arange(12,24).reshape(3,4)
a3=np.vstack((a1,a2))
print(a1)
print(a2)
print(a3)
```

```python
#horizontal splitting
a1=np.random.random((6,4))
a2=np.round(a1*100)
print(a2)
a3=np.hsplit(a2,2)
print(a3)
```

```python
#Vertical splitting
a1=np.random.random((6,4))
a2=np.round(a1*100)
print(a2)
a3=np.vsplit(a2,3)
print(a3)
```
