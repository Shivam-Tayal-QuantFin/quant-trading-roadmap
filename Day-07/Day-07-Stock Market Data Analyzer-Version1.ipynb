```python
#Project: Stock Market Data Analyzer — Version 1

import numpy as np 
import matplotlib.pyplot as plt
import math
import random
stock1=np.array(list(map(int,input("Enter the prices of stock 1 : ").split())))
stock2=np.array(list(map(int,input("Enter the prices of stock 2 : ").split())))
stock3=np.array(list(map(int,input("Enter the prices of stock 3 : ").split())))

print("----------------------------------------------")
print("Average price of stock 1 :",np.mean(stock1))
print("Average price of stock 2 :",np.mean(stock2))
print("Average price of stock 3 :",np.mean(stock3))
print("----------------------------------------------")

for i in range(1,len(stock1)):
    per_return=((stock1[i]-stock1[i-1])/stock1[i-1])*100
    print("Stock 1 Return of Day",i,"is",per_return,"%")
print("----------------------------------------------")

for i in range(1,len(stock2)):
    per_return=((stock2[i]-stock2[i-1])/stock2[i-1])*100
    print("Stock 2 Return of Day",i,"is",per_return,"%")
print("----------------------------------------------")

for i in range(1,len(stock3)):
    per_return=((stock3[i]-stock3[i-1])/stock3[i-1])*100
    print("Stock 3 Return of Day",i,"is",per_return,"%")
print("----------------------------------------------")

total_return1=((stock1[-1]-stock1[0])/stock1[0])*100
total_return2=((stock2[-1]-stock2[0])/stock2[0])*100
total_return3=((stock3[-1]-stock3[0])/stock3[0])*100
print("Total return of Stock 1 is ",total_return1,"%")
print("Total return of Stock 2 is ",total_return2,"%")
print("Total return of Stock 3 is ",total_return3,"%")
print("----------------------------------------------")
print("Lowest price of Stock 1 is ",min(stock1))
print("Lowest price of Stock 2 is ",min(stock2))
print("Lowest price of Stock 3 is ",min(stock3))
print("----------------------------------------------")
print("Highest price of Stock 1 is ",max(stock1))
print("Highest price of Stock 2 is ",max(stock2))
print("Highest price of Stock 3 is ",max(stock3))
print("----------------------------------------------")
print("Standard deviation of Stock 1 is",np.std(stock1))
print("Standard deviation of Stock 2 is",np.std(stock2))
print("Standard deviation of Stock 3 is",np.std(stock3))
print("----------------------------------------------")
if (total_return1>total_return2) & (total_return1>total_return3):
    if total_return2>total_return3:
        print("Ranking of stocks are :")
        print("Stock 1 - Rank 1")
        print("Stock 2 - Rank 2")
        print("Stock 3 - Rank 3")
    else:
        print("Ranking of stocks are :")
        print("Stock 1 - Rank 1")
        print("Stock 3 - Rank 2")
        print("Stock 2 - Rank 3")

if (total_return2>total_return3) & (total_return2>total_return1):
    if total_return1>total_return3:
        print("Ranking of stocks are :")
        print("Stock 2 - Rank 1")
        print("Stock 1 - Rank 2")
        print("Stock 3 - Rank 3")
    else:
        print("Ranking of stocks are :")
        print("Stock 2 - Rank 1")
        print("Stock 3 - Rank 2")
        print("Stock 1 - Rank 3")

if (total_return3>total_return2) & (total_return3>total_return1):
    if total_return1>total_return3:
        print("Ranking of stocks are :")
        print("Stock 3 - Rank 1")
        print("Stock 1 - Rank 2")
        print("Stock 2 - Rank 3")
    else:
        print("Ranking of stocks are :")
        print("Stock 3 - Rank 1")
        print("Stock 2 - Rank 2")
        print("Stock 1 - Rank 3")

plt.plot(stock1)
plt.plot(stock2)
plt.plot(stock3)
```
