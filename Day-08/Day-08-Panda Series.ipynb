```python
#series from lists(strings)
import pandas as pd 
country=['india','pakistan','bangladesh','nepal']
y=pd.Series(country)
print(y)
print(type(y))
```

```python
#seriies from lists(integers)
import pandas as pd
import numpy as np
runs=[12,56,89,74,23]
x=pd.Series(runs)
print(x)
print(type(x))
```

```python
#series from lists(custom index)
marks=[85,56,29,75,100]
subjects=['maths','english','science','hindi','physics']
y=pd.Series(marks,index=subjects)
print(y)
print(type(y))
```

```python
#setting a name
marks=[85,56,75,95,28]
subjects=['Maths','Physics','English','Punjabi','French']
y=pd.Series(marks,index=subjects,name='Marks of Shivam')
print(y)
print(type(y))
```

```python
#Series from dictionary 
marks={
    'maths':64,
    'english':57,
    'science':89,
    'hindi':100
}
marks_series=pd.Series(marks)
print(marks_series)
print(type(marks_series))
```

```python
#Series Attributes
marks=[85,56,75,95,28]
subjects=['Maths','Physics','English','Punjabi','French']
y=pd.Series(marks,index=subjects,name='Marks of Shivam')
print(y)
print(type(y))
#size
print(y.size)
#dtype
print(y.dtype)
#name
print(y.name)
#isunique
print(y.is_unique)
x=pd.Series([12,23,45,32,23,12])
print(x.is_unique)
#index
print(y.index)
#values
print(y.values)
```

```python
#with two column
y=pd.read_csv('bollywood.csv',index_col='movie')
print(y)
```

```python
#Series methods
#head and tails
y=pd.read_csv('bollywood.csv',index_col='movie')
print(y.head(5))
print(y.tail(5))
print(y.sample(4))

#value counts
print(y.value_counts())
```

```python
#Sort values
x=pd.read_csv('kohli_ipl.csv')
print(x)
y=x.sort_values(by='runs')
```

```python
#sort_index
y=pd.read_csv('bollywood.csv',index_col='movie')
print(y.sort_index())
print(y.sort_index(inplace=True))
```

```python
#count
x=pd.read_csv('kohli_ipl.csv')
print(x)
print(x.count())
print(x.sum())
print(x.product())
print(x.mean())
print(x.median())
print(x.std())
print(x.mode())
print(x.min())
print(x.max())
```

```python
#describe
x=pd.read_csv('kohli_ipl.csv')
print(x)
print(x.describe())
```

```python
#Series indexing
#integer indexing
x=pd.Series([12,12,14,15,25,46,57,85,79,45])
print(x[1])
#slicing
print(x[5:8])
#fancy slicing
print(x[[1,5,4,2]])
```

```python
#Editing series
#using indexing
marks=[85,56,29,75,100]
subjects=['maths','english','science','hindi','physics']
y=pd.Series(marks,index=subjects)
print(y)
y[1]=90
print(y)

y['sst']=90
print(y)
```

```python
#series with python functionalities
x=pd.read_csv('subs.csv')
print(len(x))
print(type(x))
print(dir(x))
print(sorted(x))
print(max(x))
print(min(x))
```

```python
#Type conversion
marks=[85,56,29,75,100]
subjects=['maths','english','science','hindi','physics']
y=pd.Series(marks,index=subjects)
print(y)
x=list(y)
print(x)
z=dict(y)
print(z)
```
